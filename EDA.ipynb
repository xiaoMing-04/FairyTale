# Introduction

Trong kỷ nguyên số hóa, khi công nghệ phát triển với tốc độ chóng mặt, con người ngày càng phụ thuộc vào các thiết bị thông minh để kết nối, tìm kiếm thông tin và giải trí. **Điện thoại thông minh** đã trở thành một phần không thể thiếu trong cuộc sống hàng ngày, không chỉ giúp người dùng tiếp cận thông tin nhanh chóng mà còn đóng vai trò quan trọng trong thương mại điện tử, truyền thông và quảng cáo trực tuyến. Nhờ vào khả năng kết nối liên tục, điện thoại thông minh đã thay đổi cách con người tương tác với thế giới kỹ thuật số, tạo ra một hệ sinh thái nơi dữ liệu và nội dung được tiêu thụ mọi lúc, mọi nơi.  

Với sự phát triển của nền kinh tế số, **quảng cáo trực tuyến** đã trở thành một trong những lĩnh vực quan trọng nhất đối với các doanh nghiệp và nền tảng số. Các công ty công nghệ lớn như **Google, Facebook, và TikTok** thu về hàng tỷ đô la từ quảng cáo kỹ thuật số, nơi mà nội dung được cá nhân hóa nhằm tiếp cận người dùng mục tiêu một cách hiệu quả nhất. Thống kê cho thấy **gần một nửa lưu lượng truy cập trực tuyến** đến từ thiết bị di động, tạo ra cơ hội khổng lồ cho ngành quảng cáo.  

Trong bối cảnh đó, **Tỷ lệ nhấp chuột (Click-Through Rate - CTR)** trở thành một chỉ số quan trọng giúp đánh giá mức độ hiệu quả của quảng cáo trực tuyến. CTR phản ánh tỷ lệ người dùng nhấp vào quảng cáo sau khi nhìn thấy nó, đóng vai trò quan trọng trong việc tối ưu hóa chiến dịch quảng cáo, tăng khả năng tiếp cận khách hàng và tối đa hóa lợi nhuận. Hiểu rõ về CTR và cách tối ưu hóa nó không chỉ giúp doanh nghiệp tiếp cận khách hàng tiềm năng mà còn giúp các nền tảng quảng cáo phân bổ ngân sách hiệu quả hơn, cải thiện trải nghiệm người dùng và nâng cao hiệu suất của các chiến dịch quảng cáo.

# Data Overview

## Description

Trong lĩnh vực quảng cáo trực tuyến, **Tỷ lệ nhấp chuột (Click-Through Rate - CTR)** là một chỉ số quan trọng để đánh giá hiệu quả của quảng cáo. Các hệ thống dự đoán CTR đóng vai trò cốt lõi trong **quảng cáo tìm kiếm có tài trợ (sponsored search)** và **đấu giá thời gian thực (real-time bidding - RTB)**, giúp tối ưu hóa chiến lược quảng cáo và tăng doanh thu.  

Dự án này tập trung vào xây dựng mô hình dự đoán CTR dựa trên **Avazu dataset**. Trong đó, dữ liệu Avazu cung cấp thông tin về 11 ngày hiển thị quảng cáo, tạo cơ hội để huấn luyện và kiểm thử các mô hình dự đoán hiệu quả. Mục tiêu là tìm ra chiến lược tối ưu, vượt trội so với các thuật toán phân loại tiêu chuẩn, từ đó cải thiện độ chính xác trong việc xác định khả năng người dùng sẽ nhấp vào quảng cáo.  

Kết quả của dự án có thể được mở rộng để áp dụng vào các hệ thống quảng cáo thực tế, góp phần nâng cao hiệu suất chiến dịch marketing và tối ưu hóa doanh thu từ quảng cáo trực tuyến.

## Dataset Description

### Mô tả tệp dữ liệu

- **test** - Tập dữ liệu kiểm tra. Gồm 1 ngày dữ liệu quảng cáo dùng để kiểm tra dự đoán của mô hình.

Hãy sử dụng phương thức `describe()` để xem các thông tin thống kê quan trọng của từng thuộc tính trong tập dữ liệu. Cụ thể, phương thức này sẽ cung cấp các giá trị như:  

- **Giá trị trung bình (mean):** Đại diện cho giá trị trung bình của từng thuộc tính, giúp ta có cái nhìn tổng quan về xu hướng dữ liệu.  
- **Độ lệch chuẩn (std - standard deviation):** Cho biết mức độ phân tán của dữ liệu so với giá trị trung bình. Giá trị này càng cao, dữ liệu càng có sự biến động lớn.  
- **Giá trị nhỏ nhất (min):** Giá trị thấp nhất trong tập dữ liệu cho từng thuộc tính.  
- **Các tứ phân vị (25%, 50%, 75%):** Giúp ta hiểu rõ hơn về sự phân bố của dữ liệu. Trong đó, **tứ phân vị thứ nhất (25%)** đại diện cho giá trị mà 25% dữ liệu nằm dưới đó, **tứ phân vị thứ hai (50%) hay median (trung vị)** là giá trị chính giữa của dữ liệu, và **tứ phân vị thứ ba (75%)** đại diện cho ngưỡng mà 75% dữ liệu nằm dưới đó.  
- **Giá trị lớn nhất (max):** Giá trị cao nhất của từng thuộc tính trong tập dữ liệu.  

Việc phân tích các chỉ số trên sẽ giúp ta hiểu rõ hơn về sự phân bố và xu hướng của dữ liệu, từ đó có những đánh giá và quyết định phù hợp trong quá trình xử lý và mô hình hóa dữ liệu.

### Multicollinearity

**Đa cộng tuyến (Multicollinearity)** là hiện tượng xảy ra trong mô hình hồi quy tuyến tính khi một hoặc nhiều biến độc lập (features) có mối tương quan cao với nhau. Điều này khiến mô hình khó xác định ảnh hưởng riêng lẻ của từng biến đối với biến mục tiêu, làm giảm độ tin cậy của các ước lượng hệ số hồi quy.

---

Một số dấu hiệu phổ biến của đa cộng tuyến trong mô hình:  
1. **Hệ số hồi quy biến động mạnh**: Khi thêm hoặc bớt một biến, hệ số của các biến khác thay đổi đáng kể.  
2. **Hệ số không có ý nghĩa thống kê**: Một số biến có hệ số rất nhỏ hoặc giá trị p-value cao (không có ý nghĩa thống kê) dù về lý thuyết, chúng có ảnh hưởng đến biến mục tiêu.  
3. **Chỉ số VIF (Variance Inflation Factor) cao**: Nếu một biến có **VIF > 5 hoặc 10**, nó có thể đang bị đa cộng tuyến với các biến khác.  

---

```python
corr = df.select_dtypes('number').corr()

plt.figure(figsize=(14, 10))
sns.set_style("white")

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    linecolor='white',
    vmin=-1, vmax=1,
    annot_kws={"size": 12},
    cbar_kws={"shrink": 0.8, "aspect": 30}
)

plt.title("Correlation Heatmap", fontsize=20, fontweight='bold', pad=20)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# plt.savefig('drive/MyDrive/img/img2.png')
plt.show()
```

Dựa trên biểu đồ **heatmap**, ta thấy `device_type` có tương quan mạnh với `C1` (**89%**), trong khi `C17` và `C14` có mức tương quan rất cao (**98%**). Để quyết định loại bỏ cột nào trong số các cột có tương quan mạnh này, ta sẽ dựa trên **tỷ lệ CTR** của từng cột để đánh giá mức độ ảnh hưởng của chúng đến dự đoán và giữ lại những cột mang lại giá trị thông tin cao nhất.

Chúng ta sẽ thử vẽ **boxplot** cho một số thuộc tính và nhận thấy sự xuất hiện của nhiều **outliers**. Những giá trị này có thể ảnh hưởng đến phân phối dữ liệu, làm sai lệch thống kê và tác động đến hiệu suất của mô hình dự đoán. ## Analysis of context based features

Chúng ta sẽ thực hiện **tính tỷ lệ Click-Through Rate (CTR)** cho từng giá trị của một thuộc tính bất kỳ trong tập dữ liệu. Việc tính CTR giúp ta đánh giá mức độ ảnh hưởng của từng thuộc tính đến hành vi nhấp chuột của người dùng, từ đó có thể sử dụng làm đặc trưng đầu vào cho mô hình dự đoán.  

Công thức tính CTR:  

$$
CTR = \frac{\text{Số lần nhấp (clicks)}}{\text{Tổng số lần hiển thị (impressions)}}
$$

Trong đó:  
- **Clicks**: Số lần người dùng nhấp vào quảng cáo (giá trị `click = 1`).  
- **Impressions**: Số lần quảng cáo được hiển thị (`click = 0` + `click = 1`).  

---

Đầu tiên, chúng ta nhóm dữ liệu theo từng giá trị của thuộc tính đang xét và trạng thái nhấp chuột (`click`), từ đó phân chia dữ liệu thành các nhóm dựa trên số lần hiển thị quảng cáo tương ứng với `click = 0` (không nhấp) và `click = 1` (có nhấp). Sau đó, chúng ta tính toán tần suất xuất hiện của từng giá trị trong thuộc tính đó để hiểu rõ hơn về phân bố dữ liệu. Tiếp theo, tỷ lệ CTR được tính bằng cách lấy số lần nhấp (`click = 1`) chia cho tổng số lần hiển thị của quảng cáo, bao gồm cả số lần không nhấp và có nhấp. Cuối cùng, bảng kết quả được xuất ra với đầy đủ thông tin về số lần không nhấp chuột, số lần có nhấp chuột, tần suất xuất hiện của từng giá trị thuộc tính và tỷ lệ CTR tương ứng, giúp ta đánh giá mức độ ảnh hưởng của thuộc tính đó đến hành vi nhấp chuột của người dùng.

Việc tính CTR cho từng thuộc tính giúp ta hiểu được yếu tố nào có ảnh hưởng lớn đến khả năng người dùng nhấp vào quảng cáo. Từ đó, có thể:  
- **Loại bỏ những thuộc tính không đóng góp nhiều** vào dự đoán CTR.  
- **Xử lý các thuộc tính có quá nhiều giá trị phân biệt** (như `device_id`, `device_ip`) bằng phương pháp **hashing** hoặc **tính CTR trực tiếp** thay vì one-hot encoding.  
- **Phát hiện và xử lý dữ liệu mất cân bằng**, từ đó cân bằng lại dữ liệu bằng cách lấy mẫu hoặc trọng số phù hợp trong mô hình.  

Đây là một bước quan trọng trong quá trình **khám phá dữ liệu (EDA)** và **trích xuất đặc trưng (Feature Engineering)**, giúp cải thiện hiệu suất của mô hình dự đoán CTR.

### Categorical  features

#### 1. Banner position

Vị trí hiển thị của quảng cáo trên màn hình di động là một trong những yếu tố quan trọng và cơ bản nhất ảnh hưởng đến hiệu suất của quảng cáo. Trong tập dữ liệu này, các vị trí hiển thị được chia thành 7 nhóm khác nhau. Để hiểu rõ hơn về sự phân bố số lượt nhấp chuột theo từng vị trí, chúng ta sẽ trực quan hóa dữ liệu bằng cách vẽ biểu đồ số lượt nhấp chuột tương ứng với từng vị trí hiển thị của banner quảng cáo.

```python
group_data = df.groupby(['banner_pos', 'click']).size().unstack()
group_data['Frequency'] = df['banner_pos'].value_counts(normalize=True)
ctr = group_data[1] / (group_data[0] + group_data[1])
group_data['CTR'] = ctr
display(group_data)

print('Mean of Frequency:', group_data['Frequency'].mean())
print('Mean of CTR (Click-Through-Rate):', group_data['CTR'].mean())
```

Từ bảng trên, chúng ta có thể thấy rằng phần lớn các quảng cáo được hiển thị ở vị trí 0 và 1, trong khi các vị trí từ 2 đến 7 cũng có quảng cáo nhưng với số lượng rất nhỏ. Sự chênh lệch quá lớn giữa các nhóm khiến cho các vị trí từ 2 đến 7 gần như không thể nhìn thấy rõ trên biểu đồ. Mặc dù phần lớn quảng cáo được hiển thị ở vị trí 0 và 1, nhưng tỷ lệ CTR lại thấp hơn so với vị trí 7, vốn có số lần hiển thị rất nhỏ. Điều này có thể do **hiệu ứng chọn lọc**, khi chỉ một số ít quảng cáo xuất hiện ở vị trí 7 và những quảng cáo này có thể có nội dung hấp dẫn hoặc phù hợp với nhu cầu của người dùng hơn, dẫn đến tỷ lệ nhấp chuột cao hơn đáng kể. Ngược lại, với số lần hiển thị áp đảo, các quảng cáo ở vị trí 0 và 1 phải cạnh tranh nhiều hơn để thu hút sự chú ý, khiến mức độ quan tâm của người dùng bị phân tán, từ đó kéo CTR trung bình xuống thấp hơn.

## Analysis of the data wrt to time

Chúng ta biết rằng không phải tất cả các cột đều cần thiết cho dự đoán của mình, vì vậy chúng ta nên loại bỏ các cột không cần thiết. Chúng ta loại bỏ cột giờ vì trước đó đã tạo ra cột `hour_of_day` cùng với cột `day_of_week`. Do dữ liệu của chúng ta chỉ kéo dài trong 10 ngày, nên không thể nắm bắt được xu hướng theo tháng và năm, vì vậy không có lý do gì để đưa chúng vào phân tích. Tuy nhiên, chúng ta sẽ quan tâm đến ngày trong tuần có số lần nhấp quảng cáo cao hơn, vì vậy chúng ta sẽ giữ lại cột ngày. Tương tự, việc hoạt động của người dùng thay đổi theo từng giờ là rất quan trọng đối với phân tích của chúng ta. Chúng ta đã quan sát thấy rằng cột phút và giây không đóng góp gì cho phân tích vì dữ liệu của chúng ta được thu thập theo từng giờ trong ngày, vì vậy chúng ta sẽ bỏ qua chúng. Cuối cùng, chúng ta cũng loại bỏ cột id.

<center>
    <h1>Feature Engineering</h1>
</center>

**Feature Engineering** (Kỹ thuật xây dựng đặc trưng) là một bước quan trọng trong quá trình xử lý dữ liệu và xây dựng mô hình học máy. Nó bao gồm việc tạo ra, biến đổi, hoặc chọn lọc các đặc trưng (features) từ dữ liệu thô để giúp mô hình học hiệu quả hơn. Feature Engineering có thể bao gồm nhiều kỹ thuật như chuẩn hóa dữ liệu (normalization), mã hóa dữ liệu phân loại (encoding categorical data), tạo ra các đặc trưng mới từ dữ liệu có sẵn, hoặc giảm số chiều của dữ liệu bằng các phương pháp như PCA (Principal Component Analysis). Một đặc trưng tốt có thể giúp mô hình học máy đạt độ chính xác cao hơn, giảm overfitting và cải thiện khả năng tổng quát hóa. Do đó, Feature Engineering thường đòi hỏi sự hiểu biết sâu về dữ liệu cũng như bài toán cụ thể để tìm ra những đặc trưng có ý nghĩa nhất.

### Feature Engineering trong bài toán dự đoán CTR

Trong bài toán dự đoán tỷ lệ nhấp chuột (Click-Through Rate - CTR), Feature Engineering đóng vai trò then chốt trong việc cải thiện độ chính xác của mô hình. Dữ liệu CTR thường có tính chất lớn, đa dạng và không đồng nhất, bao gồm các đặc trưng như thông tin người dùng (tuổi, giới tính, thiết bị), thông tin quảng cáo (loại quảng cáo, vị trí hiển thị, thời gian), và ngữ cảnh (trình duyệt, địa lý, lịch sử nhấp chuột).

Các kỹ thuật Feature Engineering phổ biến trong CTR prediction bao gồm:

- **Mã hóa dữ liệu phân loại (Categorical Encoding)**: Do dữ liệu CTR chứa rất nhiều biến dạng phân loại (như ID người dùng, ID quảng cáo...), cần áp dụng các phương pháp như one-hot encoding, target encoding, hoặc hashing trick để xử lý hiệu quả.
- **Tạo đặc trưng tương tác (Feature Interaction)**: Kết hợp các đặc trưng với nhau để tạo ra đặc trưng mới, ví dụ: *user_age × ad_category* có thể giúp mô hình hiểu rõ hơn mối liên hệ giữa người dùng và loại quảng cáo.
- **Trích xuất đặc trưng thời gian (Time Features)**: Biến đổi dữ liệu thời gian thành các đặc trưng như giờ trong ngày, ngày trong tuần, cuối tuần hay ngày thường, vì hành vi người dùng có thể thay đổi theo thời điểm.
- **Lịch sử hành vi người dùng (User Behavior Features)**: Tổng hợp các chỉ số như số lần nhấp, số lần hiển thị, CTR trung bình theo từng người dùng hoặc từng quảng cáo.
- **Giảm chiều dữ liệu**: Dữ liệu CTR có thể có hàng triệu đặc trưng sau khi mã hóa, do đó cần áp dụng kỹ thuật như PCA hoặc embedding để giảm chiều và tăng hiệu quả xử lý.

Việc xây dựng đặc trưng hiệu quả không chỉ giúp mô hình học được tốt hơn mà còn giảm thiểu độ nhiễu và cải thiện khả năng tổng quát hóa trong các hệ thống quảng cáo thực tế.

Sự khác biệt này chỉ ra rằng `device_id_count` thể hiện sự phân phối ổn định hơn, điều này cho thấy nó có thể đóng góp vào việc **tối ưu hoá hiệu suất quảng cáo**. Bằng cách sử dụng `device_id_count`, mô hình có thể nhận diện rõ hơn các xu hướng và hành vi của người dùng, từ đó nâng cao khả năng dự đoán **CTR**.

Điều này cho thấy `device_ip_count` cũng có khả năng đóng góp vào việc **nâng cao hiệu suất quảng cáo**, giúp mô hình dự đoán **CTR** chính xác hơn bằng cách phản ánh tốt hơn sự phổ biến và tần suất xuất hiện của địa chỉ IP trong dữ liệu.

## 4. Device ID, Device Model and User

Chúng ta hãy cùng phân tích các đặc trưng `device_id`, `device_model` và `user` để hiểu rõ hơn về cách thiết bị và người dùng được mã hóa trong dữ liệu. Việc phân tích này giúp ta xác định xem các giá trị trong những cột này mang tính phân biệt thực sự hay chỉ là các mã định danh trùng lặp hoặc bị trừu tượng hóa quá mức. Qua đó, chúng ta có thể đánh giá mức độ đóng góp của từng đặc trưng vào quá trình học của mô hình và cân nhắc việc giữ lại hay loại bỏ nếu cần thiết.

```python
# user
group_data = df.groupby(['device_id', 'click']).size().unstack()
ctr1 = group_data[1] / (group_data[1] + group_data[0])

group_data = df.groupby(['device_model', 'click']).size().unstack()
ctr2 = group_data[1] / (group_data[1] + group_data[0])

group_data = df.groupby(['user', 'click']).size().unstack()
ctr3 = group_data[1] / (group_data[1] + group_data[0])

fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 6))
axes = axes.ravel()

sns.kdeplot(ctr1, fill=True, color=sns.color_palette("magma")[4], alpha=0.8, linewidth=2, ax=axes[0])
sns.kdeplot(ctr2, fill=True, color=sns.color_palette("coolwarm")[4], alpha=0.8, linewidth=2, ax=axes[1])
sns.kdeplot(ctr3, fill=True, color=sns.color_palette("viridis")[4], alpha=0.8, linewidth=2, ax=axes[2])

axes[0].set_title("CTR device_id", fontsize=14, fontweight='bold', color='darkorange')
axes[1].set_title("CTR device_model", fontsize=14, fontweight='bold', color='darkblue')
axes[2].set_title("CTR user", fontsize=14, fontweight='bold', color='darkblue')
axes[3].axis('off')

for ax in axes:
    ax.set_xlabel("CTR", fontsize=12, fontweight='bold', color='darkred')
    ax.set_ylabel("Density", fontsize=12, fontweight='bold', color='darkred')
    ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Có khả năng mỗi `user` gần như gắn liền với một `device_id` duy nhất, hoặc chúng phản ánh cùng một thực thể theo cách khác nhau.
